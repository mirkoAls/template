```python
# Este notebook reutiliza la lógica mantenida en seach template.py.
# Cambia esta ruta antes de ejecutar la celda.
from importlib.util import module_from_spec, spec_from_file_location
from pathlib import Path

BASE_DIR = Path(r"D:\Work web-Plantillas\plantilla")
SCRIPT_PATH = Path.cwd() / "seach template.py"

spec = spec_from_file_location("template_search", SCRIPT_PATH)
if spec is None or spec.loader is None:
    raise RuntimeError(f"No se pudo cargar {SCRIPT_PATH}")
template_search = module_from_spec(spec)
spec.loader.exec_module(template_search)

exit_code = template_search.main(BASE_DIR)
if exit_code:
    raise RuntimeError(f"La búsqueda terminó con código {exit_code}")
```
